```python

input("🔘 Press Enter to start your adventure...")

# Get name
user_input = input("Hi! What is your name?: ")
words = user_input.split()
name = words[-1].capitalize()
print("Welcome, " + name + "!")

# Get age
user_input = input("What is your age?: ")
words = user_input.split()
age = words[-1]
age = ''.join(filter(str.isdigit, age))  # Clean any extra text
if age.isdigit():
    age = int(age)
else:
    age = 0  # Unknown age

# Get favorite color
user_input = input("What's your favorite color?: ")
words = user_input.split()
color = words[-1].capitalize()

print("\nA " + str(age) + "-year-old adventurer in " + color + "... your journey begins now, " + name + ".")

# First choice
print("\nYou find yourself at a crossroads.")
print("To the left: a dark forest.")
print("To the right: a glowing cave.")
print("Straight ahead: a misty mountain.")

path = input("Do you go to the forest, cave, or mountain? ").lower()

# === Forest Path ===
if "forest" in path:
    print("\n🌲 You step into the forest. The air is thick with magic.")
    print("A talking fox appears and says, 'Only the clever survive here.'")
    choice = input("Do you follow the fox or explore alone? ").lower()

    if "follow" in choice:
        print("\n🦊 The fox leads you to a hidden glade where treasure lies buried.")
        print("You have found the treasure of the forest!")
        ending = "treasure"
    else:
        print("\nYou get lost in the woods. The shadows close in...")
        print("Everything fades to black.")
        ending = "death"

# === Cave Path ===
elif "cave" in path:
    print("\n🕯️ The cave walls shimmer with ancient light.")
    print("A voice whispers: 'Touch the stone, and learn your fate.'")
    choice = input("Do you touch the stone or walk deeper into the cave? ").lower()

    if "stone" in choice:
        print("\n💫 The world spins... you open your eyes in bed.")
        print("It was all a dream.")
        ending = "dream"
    else:
        print("\nYou walk deeper and fall into a pit.")
        print("Darkness surrounds you. You will learn from this fall.")
        ending = "death"

# === Mountain Path ===
elif "mountain" in path:
    print("\n⛰️ The mist is cold, and the path is steep.")
    print("At the peak stands a sword in a stone and a sleeping dragon.")
    choice = input("Do you take the sword or sneak past the dragon? ").lower()

    if "sword" in choice:
        print("\n🔥 The dragon awakens and attacks!")
        print("You are burned... but legends say you may return wiser.")
        ending = "death"
    else:
        print("\nYou sneak past and find a chest of gold behind the dragon's nest!")
        print("You have found the mountain’s treasure!")
        ending = "treasure"

# === Unknown Path ===
else:
    print("\nYou wander with no direction. The world shifts...")
    print("You open your eyes. It was all a dream.")
    ending = "dream"

# === Endings ===
print("\n🎉 Ending Reached:")
if ending == "dream":
    print("You wake up in your bed, the adventure fading like mist. Was it real?")
elif ending == "treasure":
    print("Victory! You found the treasure and will be remembered in song and story.")
elif ending == "death":
    print("You died... but your journey taught you something. You will do better next time.")

print("\nThanks for playing, " + name + "!")
```
